Testing notebook for interpreting everything as a saved scenario and delegating everywhere to scenario methods.


## From Excel


```python

# Check the environment is properly configured.
from example_helpers import setup_notebook
setup_notebook()
```

```python
# What is now called SavedScenarios will become Scenarios.
from pyetm.models.saved_scenarios import SavedScenarios

scenarios = SavedScenarios.from_excel("../examples/inputs/example_input_excel.xlsx")
```

```python
# Metadata
for scenario in scenarios:
    print(f"Title: {scenario.title}")
    print(f"ID: {scenario.id}")
    print(f"Area: {scenario.area_code}")
    print(f"End year: {scenario.end_year}")
    print(f"Version: {scenario.version}")
    print(f"Template: {scenario.template}")
    print("")
```

```python
# Accessing the session data from the saved scenarios
for scenario in scenarios:
    print(f"Keep combatible: {scenario.keep_compatible}")
    print(f"ID: {scenario.scenario_id}")
    print(f"Inputs: {scenario.inputs.to_dataframe()}")
    print("")
```
