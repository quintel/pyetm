Testing notebook for loading mixed Scenario and Session types from Excel.

This notebook demonstrates the simplified API where `Scenarios.from_excel()` loads both SavedScenario (from MyETM) and Session (from ETEngine) instances.

## From Excel


```python

# Check the environment is properly configured.
from example_helpers import setup_notebook
setup_notebook()
```

```python
from pyetm.models.scenarios import Scenarios

# Load all scenarios (both SavedScenarios and Sessions) from the Excel file
mixed_scenarios = Scenarios.from_excel("../examples/inputs/example_input_excel.xlsx")
```

```python
# Iterate over all scenarios (both Scenario and Session instances)
# The delegation pattern means both types have the same interface
for scenario in mixed_scenarios:
    print(f"Title: {scenario.title}")
    print(f"ID: {scenario.id}")
    print(f"Area: {scenario.area_code}")
    print(f"End year: {scenario.end_year}")
    print(f"Version: {scenario.version}")
    print(f"Template: {scenario.template}")
    print("")
```

```python
# Access the underlying Session instances from all items
# For Scenario instances, this unwraps the session; for Session instances, returns them directly
sessions = mixed_scenarios.sessions

for session in sessions:
    print(f"Title: {session.title}")
    print(f"ID: {session.id}")
    print(f"Area: {session.area_code}")
    print(f"End year: {session.end_year}")
    print(f"Version: {session.version}")
    print(f"Template: {session.template}")
    print("")
```

```python
# Accessing delegated properties works on all types
for scenario in mixed_scenarios:
    print(f"Keep compatible: {scenario.keep_compatible}")
    print(f"ID: {scenario.id}")
    print(f"Type: {'SavedScenario' if hasattr(scenario, 'scenario_id') and hasattr(scenario, 'session') else 'Session'}")
    # Uncomment to see inputs:
    # print(f"Inputs: {scenario.inputs.to_dataframe()}")
    print("")
```
