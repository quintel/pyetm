Notebook to show Pyetm from a more dev perspective:

```python
# Check the environment is properly configured.
from example_helpers import setup_notebook
setup_notebook(debug=True)
```

```python
from pyetm.models.scenarios import Scenarios

loaded = Scenarios.load_many([1359192, 1359193, 1359194])
print(len(loaded))

create_params = [
    {"area_code": "nl", "end_year": 2050},
    {"area_code": "nl", "end_year": 2040},
]

created = Scenarios.create_many(create_params)


scenarios = Scenarios()
scenarios.extend(loaded)
scenarios.extend(created)

# Metadata
for scenario in scenarios:
    print(f"Title: {scenario.title}")
    print(f"ID: {scenario.id}")
    print(f"Area: {scenario.area_code}")
    print(f"End year: {scenario.end_year}")
    print(f"Version: {scenario.version}")
    print(f"Source: {scenario.source}")
    print(f"Metadata: {scenario.metadata}")
    print("")
```

```python
# Inputs
for scenario in scenarios:
    inputs = scenario.inputs.to_dataframe(columns=["unit", "key", "user", "default", "min", "max"]).head(15)
    print(inputs)
    print("")
```

```python
# Couplings
for scenario in scenarios:
    couplings = scenario.couplings
    print(couplings)
    print("")
```

```python
# Sortables
for scenario in scenarios:
    sortables = scenario.sortables.to_dataframe()
    print(sortables)
    print("")
```

```python
# Custom Curves
for scenario in scenarios:
    curves = scenario.custom_curves.to_dataframe().head(20)
    print(curves)
    print("")
```

```python
# Queries
for scenario in scenarios:
    print(scenario.results())
    print("")
```

```python
# Warnings - did anything go wrong?
for scenario in scenarios:
    print(scenario.show_all_warnings())
    print()
```
