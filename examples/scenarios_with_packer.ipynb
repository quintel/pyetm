This notebook demonstrates pyetm's programmatic scenario management and data export functionalities, focusing on bulk operations and data manipulation using Python code rather than Excel interfaces.

Check the environment setup.

```python
# Check the environment is properly configured.
from example_helpers import setup_notebook
setup_notebook()
```

Load and create scenarios.

```python
from pyetm.models.sessions import Sessions

loaded = Sessions.load_many([1359192, 1359193, 1359194])
print(len(loaded))

create_params = [
    {"area_code": "nl", "end_year": 2050},
    {"area_code": "nl", "end_year": 2040},
]

created = Sessions.create_many(create_params)


scenarios = Sessions()
scenarios.extend(loaded)
scenarios.extend(created)
```

Working with the packer - all models are converted to dataframes for easy export to other data wrangling formats or Excel.

Add scenarios to the packer.

```python
from pyetm.models.scenario_packer import ScenarioPacker

packer = ScenarioPacker()
packer.add(*scenarios)
```

Show scenario inputs.

```python
packer.inputs(columns=["user", "default", "min", "max", "permitted_values"]).head(15)
```

Show custom curves.

```python
packer.custom_curves().head(20)
```

Show sortables.

```python
packer.sortables().head(20)
```

Show exports.

```python
packer.exports().head(15)
```

Export scenarios to Excel.

```python
# The flag include_input_details includes min max and default for inputs in a separate sheet INPUT_DETAILS
packer.to_excel(include_inputs=True, include_input_details=True, include_sortables=True, path="../examples/outputs/scenarios.xlsx")
```
